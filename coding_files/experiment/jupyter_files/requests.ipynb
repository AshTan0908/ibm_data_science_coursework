```python
import requests as r
ur = 'https://www.ibm.com'
req = r.get(ur)
print(req)
print(req.request.body)
header=req.headers
print(header)
date = header['date']
ctype = header['Content-Type']
print(date)
print(ctype)
print(req.encoding)
print(req.text[0:100])
```

```python
import requests
url='https://httpbin.org/get'
payload={'name':'John', 'ID':'123'}
r = requests.get(url, params=payload)
print(r.request.body)
print(r.url)
print(r.text)
print(r.json())

urlp='https://httpbin.org/post'
p=requests.post(urlp, payload)
print(p.url)
```
