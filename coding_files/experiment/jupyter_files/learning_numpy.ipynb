```python
import numpy as np
l = [0, 1, 2, 3, 4, 5]
a = np.array(l)

type(a) # <class 'numpy.ndarray'>
print(a)
x = a.dtype # type of array elements
print(x)
print(a.size) # basically len() for arrays
print(a.ndim) # dimensions of array
print(a.shape) # length of each dimension
```
